```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python

n = np.arange(-5, 11)

delta = (n == 0).astype(float)

step = (n >= 0).astype(float)

ramp = n * step

expo = 0.8**n * step
```

```python

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["Unit Impulse", "Unit Step", "Unit Ramp", "Exponential"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

### **How many numbers are needed to store one minute of mono audio?**

**Answer:**

The number of samples required is calculated using:

N = Sampling Rate × Duration

Given:
- Sampling Rate = 44,100 Hz
- Duration = 60 seconds

N = 44,100 × 60

N = 2,646,000 samples.

Therefore, 2,646,000 samples are needed to store one minute of mono audio at a sampling rate of 44,100 Hz.



**Question 2: Why is a stem plot more appropriate than a continuous line for (x(n))?**

Answer:

A stem plot is more appropriate for discrete-time signals because it displays individual samples at specific time instants. Each sample is represented by a vertical line and a marker, making it easy to identify the amplitude of each sample.

Unlike a continuous line plot, which connects points continuously, a stem plot clearly shows that discrete-time signals exist only at specific sampling instants.



**Question 3:** For \(a = 1.1\), the exponential grows. Which property of a system would such a signal violate if it were an impulse response?

**Answer:**

It would violate the **BIBO (Bounded-Input, Bounded-Output) stability** property. This is because when \(a = 1.1\), the exponential signal grows without bound. If this signal were an impulse response, the system would not be BIBO stable. Therefore, the system would be unstable.
